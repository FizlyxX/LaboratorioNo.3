**Importación de Librerías y Carga de Datasets**

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Definir las rutas exactas de los archivos en GitHub (Raw)
url_customer = "https://raw.githubusercontent.com/FizlyxX/LaboratorioNo.3/refs/heads/main/Customer.csv"
url_invoice = "https://raw.githubusercontent.com/FizlyxX/LaboratorioNo.3/refs/heads/main/Invoice.csv"

# Cargar los datasets especificando la codificación adecuada
df_customer = pd.read_csv(url_customer, encoding="latin1")
df_invoice = pd.read_csv(url_invoice, encoding="latin1")

# Mostrar las primeras filas de cada dataset para verificar la carga
print("--- Primeras filas de Customer ---")
display(df_customer.head())

print("\n--- Primeras filas de Invoice ---")
display(df_invoice.head())
```

**Verificación de Nulos y Duplicados**

**Análisis:**
Se inspeccionan ambas tablas para identificar la presencia de filas duplicadas y el conteo de valores faltantes/nulos por columna. Esto permite planificar la limpieza antes de realizar agregaciones o cruces de datos.

```python
print("=== VERIFICACIÓN EN CUSTOMER ===")
print(f"Filas duplicadas en Customer: {df_customer.duplicated().sum()}")
print("\nMuestra de valores nulos por columna en Customer:")
print(df_customer.isnull().sum())

# Muestra de filas con valores nulos en Customer
print("\nMuestra de registros con nulos en Customer (Company):")
display(df_customer[df_customer["Company"].isnull()].head(10))

print("\n" + "=" * 40 + "\n")

print("=== VERIFICACIÓN EN INVOICE ===")
print(f"Filas duplicadas en Invoice: {df_invoice.duplicated().sum()}")
print("\nMuestra de valores nulos por columna en Invoice:")
print(df_invoice.isnull().sum())

# Muestra de filas con valores nulos en Invoice
print("\nMuestra de registros con nulos en Invoice (BillingState):")
display(df_invoice[df_invoice["BillingState"].isnull()].head(10))
```

**Revisión y conversión de tipos de datos**

**Pregunta:** ¿La columna InvoiceDate se cargó como texto o como fecha? Si es texto, conviértala con pd.to_datetime().

**Respuesta:** Al verificar con df_invoice['InvoiceDate'].dtype, la columna se carga inicialmente como objeto (object / texto). Se procede a convertirla al tipo datetime64[ns] para poder realizar operaciones temporales.

```python
# Muestra del tipo de dato original
print(
    f"Tipo de dato original de InvoiceDate: {df_invoice['InvoiceDate'].dtype}"
)

# Conversión a datetime
df_invoice["InvoiceDate"] = pd.to_datetime(df_invoice["InvoiceDate"])

# Muestra del nuevo tipo de dato y verificación de las primeras filas
print(f"Nuevo tipo de dato de InvoiceDate: {df_invoice['InvoiceDate'].dtype}")
print("\nMuestra del dataset Invoice con la fecha convertida:")
display(df_invoice[["InvoiceId", "InvoiceDate", "Total"]].head())
```

**Top 10 países con más ingresos y porcentaje del primer país**

**Pregunta:** ¿Cuáles son los 10 países (BillingCountry) que más ingresos generan? Calcule también qué porcentaje del total representa el país que va primero.

**Respuesta:** Se agrupan los ingresos por país de facturación, se ordenan de mayor a menor y se calcula el porcentaje de participación del país con mayores ventas con respecto al total global de ingresos.

```python
# Calcular ingresos totales por país
ingresos_por_pais = (
    df_invoice.groupby("BillingCountry")["Total"]
    .sum()
    .sort_values(ascending=False)
)

# Top 10 países
top_10_paises = ingresos_por_pais.head(10)

# Porcentaje del primer país
ingreso_total_global = df_invoice["Total"].sum()
primer_pais_nombre = top_10_paises.index[0]
primer_pais_ingreso = top_10_paises.iloc[0]
porcentaje_primer_pais = (primer_pais_ingreso / ingreso_total_global) * 100

print("--- Top 10 Países por Ingresos Total ---")
display(top_10_paises.to_frame(name="Ingresos_Totales"))

print(f"\nEl país que más ingresos genera es: {primer_pais_nombre}")
print(f"Monto generado: ${primer_pais_ingreso:,.2f}")
print(
    f"Porcentaje que representa del total global: {porcentaje_primer_pais:.2f}%"
)
```

**Ticket promedio global y comparación (EE. UU. vs. Francia)**

**Pregunta:** ¿Cuál es el ticket promedio global? ¿Y cómo se compara el ticket promedio de Estados Unidos contra el de Francia?

**Respuesta:** El ticket promedio se calcula como el promedio de la columna Total de cada factura. Se extrae la media global y se comparan los promedios específicos de facturación en USA y Francia.

```python
# Ticket promedio global
ticket_promedio_global = df_invoice["Total"].mean()

# Ticket promedio por país
ticket_usa = df_invoice[df_invoice["BillingCountry"] == "USA"]["Total"].mean()
ticket_francia = df_invoice[df_invoice["BillingCountry"] == "France"][
    "Total"
].mean()

print(f"Ticket Promedio Global: ${ticket_promedio_global:.2f}")
print(f"Ticket Promedio en USA: ${ticket_usa:.2f}")
print(f"Ticket Promedio en Francia: ${ticket_francia:.2f}")

diferencia = ticket_usa - ticket_francia
if diferencia > 0:
  print(f"El ticket promedio de USA es ${abs(diferencia):.2f} mayor que el de Francia.")
else:
  print(f"El ticket promedio de Francia es ${abs(diferencia):.2f} mayor que el de USA.")
```

**Tabla resumen con .agg() por país**

Instrucción: Usando .agg(), construya una sola tabla que muestre, por país: ingresos totales, número de facturas, y ticket promedio, en una sola instrucción, con columnas nombradas.

```python
# Construcción del resumen utilizando .agg() con asignación de nombres de columna
resumen_pais = df_invoice.groupby("BillingCountry").agg(
    Ingresos_Totales=("Total", "sum"),
    Numero_Facturas=("InvoiceId", "count"),
    Ticket_Promedio=("Total", "mean"),
)

# Ordenar por Ingresos Totales descendentemente para mejor visualización
resumen_pais = resumen_pais.sort_values(
    by="Ingresos_Totales", ascending=False
)

print("--- Muestra de la Tabla Resumen por País ---")
display(resumen_pais.head(10))
```

**Merge y verificación de inconsistencias de país**

**Pregunta:** Cada cliente en Customer tiene un país registrado (Country), y cada factura en Invoice tiene un país de facturación (BillingCountry). Haga un merge de las dos tablas y verifique: ¿hay algún caso donde el país del cliente no coincide con el país de la factura?

**Análisis:** Se realiza un cruce mediante CustomerId para verificar si existen facturas cuyo BillingCountry difiera del Country registrado en el perfil del cliente.

```python
# Merge de las dos tablas
df_merged = pd.merge(
    df_invoice,
    df_customer,
    on="CustomerId",
    how="inner",
    suffixes=("_Invoice", "_Customer"),
)

# Filtro para verificar diferencias de país
diferencias_pais = df_merged[
    df_merged["Country"] != df_merged["BillingCountry"]
]

print(
    f"Número de casos donde el país del cliente no coincide con el de la factura: {len(diferencias_pais)}"
)

if len(diferencias_pais) > 0:
  print("\nMuestra de las discrepancias encontradas:")
  display(
      diferencias_pais[[
          "InvoiceId",
          "CustomerId",
          "Country",
          "BillingCountry",
      ]].head()
  )
else:
  print(
      "\nEn todos los registros, el país del cliente coincide con el país de la factura."
  )
```

**Verificación de clientes sin facturas**

**Pregunta:** ¿Hay algún cliente en Customer que nunca haya generado una factura? Resuélvalo con el tipo de unión adecuado.

**Respuesta:** Se realiza un Left Join desde Customer hacia Invoice (o Right Join según la perspectiva). Los clientes que no poseen facturas asociadas mostrarán valores nulos (NaN) en los campos correspondientes a la tabla Invoice.

```python
# Unión Left Join conservando todos los clientes
df_clientes_facturas = pd.merge(
    df_customer, df_invoice, on="CustomerId", how="left"
)

# Identificar clientes con InvoiceId nulo
clientes_sin_factura = df_clientes_facturas[
    df_clientes_facturas["InvoiceId"].isnull()
]

print(f"Cantidad de clientes sin facturas: {len(clientes_sin_factura)}")

if len(clientes_sin_factura) > 0:
  print("\nMuestra de clientes que nunca han facturado:")
  display(
      clientes_sin_factura[[
          "CustomerId",
          "FirstName",
          "LastName",
          "Email",
      ]].drop_duplicates()
  )
else:
  print("\nTodos los clientes registrados tienen al menos una factura asociada.")
```

**Top 10 de clientes por gasto total**

**Instrucción:** Usando el merge, construya el top 10 de clientes por gasto total, mostrando el nombre completo del cliente (no solo el CustomerId).

```python
# Crear columna con Nombre Completo
df_merged["FullName"] = df_merged["FirstName"] + " " + df_merged["LastName"]

# Agrupar por Cliente y sumar gasto total
top_clientes = (
    df_merged.groupby(["CustomerId", "FullName"])["Total"]
    .sum()
    .reset_index()
    .rename(columns={"Total": "GastoTotal"})
    .sort_values(by="GastoTotal", ascending=False)
)

print("--- Top 10 Clientes por Gasto Total ---")
display(top_clientes.head(10))
```

**Tabla pivote (Año vs. País)**

**Instrucción:** Construya una tabla pivote con el año de la factura en las filas y el país en las columnas, mostrando la suma de Total.

```python
# Extraer el año de InvoiceDate
df_invoice["Year"] = df_invoice["InvoiceDate"].dt.year

# Crear tabla pivote
tabla_pivote = pd.pivot_table(
    df_invoice,
    values="Total",
    index="Year",
    columns="BillingCountry",
    aggfunc="sum",
    fill_value=0,
)

print("--- Tabla Pivote: Ventas por Año y País (Muestra de columnas) ---")
# Se muestran las primeras columnas para visualización clara en Colab
display(tabla_pivote.iloc[:, :14])
```

**Tasa de recompra de clientes**

**Pregunta:** ¿Qué porcentaje de los clientes tiene más de una factura?

**Respuesta:** Se calcula dividiendo el número de clientes que tienen 2 o más facturas entre la cantidad total de clientes únicos registrados en la base de datos.

```python
# Conteo de facturas por cliente
facturas_por_cliente = df_invoice.groupby("CustomerId")["InvoiceId"].count()

# Total de clientes en la base de datos
total_clientes_unicos = df_customer["CustomerId"].nunique()

# Clientes con más de una factura (recompra)
clientes_recompra = (facturas_por_cliente > 1).sum()

# Cálculo del porcentaje
tasa_recompra = (clientes_recompra / total_clientes_unicos) * 100

print(f"Total de clientes únicos: {total_clientes_unicos}")
print(f"Clientes con más de una factura: {clientes_recompra}")
print(f"Tasa de recompra: {tasa_recompra:.2f}%")
```
